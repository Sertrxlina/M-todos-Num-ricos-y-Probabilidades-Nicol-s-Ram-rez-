```python
#Tarea 4.6

#en el intervalo de masa alta hay muy pocos halos
#entonces no se puede usar el TCL como si nada pq la gaussiana se cae
#simular con poisson sería mejor

import numpy as np
import matplotlib.pyplot as plt

# el modelo dice q en promedio hay 3 halos en ese rango
#(es bueno poner 3 pq es un numero chico y se nota el efecto)
media = 3

#simulo las mediciones como poisson porque son eventos raros
N = 10000
mediciones = np.random.poisson(media, N)

# saco la media y la desvest de lo simulado
#(pa comparar despues con la gaussiana)
media_medida = np.mean(mediciones)
desv_medida = np.std(mediciones)

print("media:", media_medida)
print("desviacion:", desv_medida)

# grafico el histograma de lo q sale de verdad (poisson)
plt.hist(mediciones, bins=np.arange(-0.5, 12.5, 1),
         density=True, alpha=0.7, color='orange', label='datos (poisson)')

# ahora la gaussiana que usaria el paper que estaría mal
x = np.linspace(-2, 10, 200)
gaussiana = (1 / (desv_medida * np.sqrt(2*np.pi))) * \
            np.exp(-(x-media_medida)**2 / (2*desv_medida**2))

plt.plot(x, gaussiana, linewidth=2, color='red', label='gaussiana (TCL)')

# detalles del grafico
plt.xlabel("numero de halos")
plt.ylabel("probabilidad")
plt.title("halos masivos: poisson vs gaussiana")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

#la gaussiana predice q los valores altos (ej 8 o 9 halos) son CASI IMPOSIBLES
#pero la poisson le da una probabilidad chica pero NO nula
```
