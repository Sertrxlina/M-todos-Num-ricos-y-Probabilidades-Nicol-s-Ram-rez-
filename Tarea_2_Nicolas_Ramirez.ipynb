```python
#INCOMPLETA

import numpy as np
import math

# El valor de x que nos entrega el ejercicio
x = 10**(-6)
print("Valor de x:", x)
print("------------------------------------")

# PARTE 1: CALCULAR f(x)
# La formula es: f(x) = (1 - cos(x)) / sin(x)

# Primera operacion: calcular cos(x)
cos_x = np.cos(x)

# Segunda operacion: restar 1 - cos(x)
numerador_f = 1 - cos_x

# Tercera operacion: calcular sin(x)
sen_x = np.sin(x)

# Cuarta operacion: dividir el numerador por el seno
f = numerador_f / sen_x

print("Resultado de f(x):", f)
print("------------------------------------")

#PARTE 2: CALCULAR g(x)

#La formula es: g(x) = 2 * sin(x/2)^2 / sin(x)

mitad_x = x / 2
sen_mitad = np.sin(mitad_x)
sen_cuadrado = sen_mitad**2
numerador_g = 2 * sen_cuadrado
sen_x_g = np.sin(x)
g = numerador_g / sen_x_g

print("Resultado de g(x):", g)
print("------------------------------------")

# PARTE 3: NUMERO DE CONDICIONAMIENTO DE f(x)

# El numero de condicionamiento relativo se calcula como: k = | x * f'(x) / f(x) |
# Para esta funcion: f(x) = tan(x/2) Por lo tanto: k = | x / sin(x) |

cond_f = abs(x / np.sin(x))

print("Numero de condicionamiento de f(x):", cond_f)
print("------------------------------------")

#PARTE 4: NUMERO DE CONDICIONAMIENTO DE g(x)
# Como f(x) y g(x) son matematicamente equivalentes,
# tienen el mismo numero de condicionamiento.

cond_g = abs(x / np.sin(x))

print("Numero de condicionamiento de g(x):", cond_g)
print("------------------------------------")

#PARTE 5: CONDICIONAMIENTO DE CADA OPERACION DE f


# Operacion 1: cos(x)
# Para cos(x), el condicionamiento relativo es:
# k = |x * tan(x)|

cond_cos = abs(x * np.tan(x))

# Operacion 2: resta 1 - cos(x) Esta resta puede tener cancelacion numerica.
#Se usa la formula: k = (|a| + |b|) / |a - b|

cond_resta = (abs(1) + abs(cos_x)) / abs(numerador_f)

#Operacion 3: sin(x)
#k = |x * cot(x)|
#cot(x) = 1 / tan(x)

cond_sen = abs(x / np.tan(x))

#Operacion 4: division
# Para una division a/b, el condicionamiento
#relativo respecto a ambos valores es 2.

cond_div = 2


print("CONDICIONAMIENTO DE LAS OPERACIONES DE f(x)")
print("cos(x):", cond_cos)
print("1 - cos(x):", cond_resta)
print("sin(x):", cond_sen)
print("Division:", cond_div)
print("------------------------------------")


```
