```python
import math

e_real = math.exp(1) #valor real de e usando la funcion que nos dieron

e_aprox = 0 #aqui se irá guardando la suma
n = 0 # se parte desde el 0
error = 1 #se coloca un error grande ya que asi entraria al while

while error >= 0.01: #indica que se vayan sumando términos hasta que el error sea menor que 0.01

  termino = 1 /math.factorial(n) #cálculo del término 1/n!
  e_aprox = e_aprox + termino #lo sumamos a la aproximacion de e
  error = abs(e_real - e_aprox)

  n = n + 1 #pasamos al sgt valor de n

print("Valor aprox de e:", e_aprox)
print("Valor real de e:", e_real)
print("Error:", error)
```
