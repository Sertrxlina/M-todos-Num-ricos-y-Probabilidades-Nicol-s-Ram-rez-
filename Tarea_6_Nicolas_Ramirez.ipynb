Intente escribir su propia versión del método de Newton, haciéndola lo más
rápida posible. Luego compare el tiempo que tarda en resolver el ejemplo de
clase (el cero de la función $f(x) = x - \cos(x)$) con el tiempo que tarda el
método de Newton que se encuentra en scipy.

```python
import numpy as np
import matplotlib.pyplot as plt
import timeit
from math import cos, sin
from scipy import optimize


#funcion a la que le queremos hallar el cero: f(x) = x - cos(x)
def f(x):
    return x - np.cos(x)
#derivada de f, la necesitamos para Newton: f'(x) = 1 + sin(x)
def df(x):
    return 1 + np.sin(x)

#excepcion propia por si nos pasamos del numero de iteraciones
class MaxIterations(Exception):
    pass

def mi_newton(p0, epsilon, N=1000000): #implementacion del metodo
                                       #de Newton-Raphson

    for i in range(N):
        #Formula de Newton usando f y df
        #basicamente p0 - f(p0)/f'(p0))
        p = p0 - (p0 - cos(p0)) / (1 + sin(p0))

        #criterio de parada: error relativo entre iteraciones
        if abs(p - p0) / abs(p) < epsilon:
            return p, i + 1  #raiz y numero de iteraciones usadas

        p0 = p

    # Si no converge, se "avisa"
    raise MaxIterations("No se encontro el cero")


#probando un solo pto inicial
x0 = 1
epsilon = 0.00001

#llamo al metodo
raiz1, iter1 = mi_newton(x0, epsilon)

#metodo de scipy (newton) para comparar resultados
sol = optimize.root_scalar(f, x0=x0, fprime=df, method='newton')

print("Mi metodo:")
print("Raiz:", raiz1)
print("Iteraciones:", iter1)

print("\nMetodo de scipy:")
print("Raiz:", sol.root)
print("Iteraciones:", sol.iterations)
print("Llamadas a la funcion:", sol.function_calls)


#Ahora medimos tiempos de ejecucion
n = 10000  # numero de ejecuciones por medicion

# timeit.repeat corre varias veces y nos quedamos con el mejor tiempo
# luego dividimos entre n para tener tiempo por llamada
tiempo1 = min(timeit.repeat(
    lambda: mi_newton(x0, epsilon),
    number=n,
    repeat=5
)) / n

tiempo2 = min(timeit.repeat(
    lambda: optimize.root_scalar(f, x0=x0, fprime=df, method='newton'),
    number=n,
    repeat=5
)) / n

print("Tiempo de mi metodo:", tiempo1, "segundos")
print("Tiempo de scipy:", tiempo2, "segundos")

#comparamos cual fue mas rapido
if tiempo1 < tiempo2:
    print("\nMi implementacion fue mas rapida.")
else:
    print("\nScipy fue mas rapido.")

print("\nRelacion de tiempos:", tiempo1 / tiempo2)
```
